```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

### Q1: Load the Dataset
Load the dataset titanic.csv using appropriate data structure. Do not use any other library like csv, pandas, numpy, etc.

Print the total number of rows (including the header) in the dataset.



```python
# TODO: ADD YOUR CODE HERE
with open('titanic.csv', 'r') as file:
    data = file.read().splitlines()
    print(f'Total rows in dataset: {len(data)}')
```

### Q2: Print the Column Names
Print all the column names from the dataset.

```python
# TODO: ADD YOUR CODE HERE
col_names = data.pop(0).split(',')
print(f'Column Names: {col_names}')
```

### Q3: Print the First 25 Rows
Print the first 25 rows of the dataset, excluding the header.



```python
# TODO: ADD YOUR CODE HERE
data = [item.split(',') for item in data]
for i in range(25):
    print(data[i])
```

### Q4: Filter the Dataset
Remove all unnecessary columns and retain only the relevant columns: Survived, Pclass, Sex, and Age. Print the new header and the first 10 rows of the filtered dataset.

```python
# TODO: ADD YOUR CODE HERE
print("Filtered Header:\n['Survived', 'Pclass', 'Sex', 'Age']\n")

filtered = []
for person in data:
    filtered.append([person[1], person[2], person[5], person[6]])

for i in range(10):
    print(filtered[i])
```

### Q5: Process Missing and Categorical Data


### Q5a: Convert the Sex Column to Numeric
Replace the values in the Sex column:
*   Convert "male" to 0.
*   Convert "female" to 1.

Print the first 10 rows of the dataset after this conversion.



```python
# TODO: ADD YOUR CODE HERE
print("First 10 rows after converting 'Sex' to numeric:\n")

for person in filtered:
    if person[2] == 'male':
        person[2] = 0
    elif person[2] == 'female':
        person[2] = 1

for i in range(10):
    print(filtered[i])
```

### Q5b: Handle Missing Age Values
Remove the rows with empty Age values. Print the first 10 rows of the dataset after handling missing values.

```python
# TODO: ADD YOUR CODE HERE
print(f"Number of rows before dropping missing 'Age' values: {len(filtered)}")

for person in filtered:
    if person[3] == '':
        filtered.remove(person)

print(f"Number of rows before dropping missing 'Age' values: {len(filtered)}")
print("First 10 rows after handling missing 'Age' values: ")

for i in range(10):
    print(filtered[i])
```

### Q6: Split Features and Labels
Separate the dataset into features (Pclass, Sex, Age) and labels (Survived). Print the first 10 rows of each.

```python
# TODO: ADD YOUR CODE HERE
features = [item[1:4] for item in filtered]
labels = [item[0] for item in filtered]

print("Features: ")
for i in range(10):
    print(features[i])

print("\nLabels: ")
for i in range(10):
    print(labels[i])
```

### Q7: Split Data for Training and Testing
Split the features and labels into training and testing sets (80%-20%). Print the number of rows in the training and testing sets.

```python
# TODO: ADD YOUR CODE HERE
x_train, x_test, y_train, y_test = train_test_split(
    features, labels,
    test_size=0.2,
    random_state=42
)

print(f'Training set size: {len(x_train)}')
print(f'Testing set size: {len(x_test)}')
```

### Q8: Train a Decision Tree Classifier
Train a DecisionTreeClassifier on the training data. Print a message indicating the training is complete.

```python
# TODO: ADD YOUR CODE HERE
model = DecisionTreeClassifier(random_state=42)
model.fit(x_train, y_train)

print('Model training complete.')
```

### Q9: Evaluate the Model
Use the test set to calculate model's accuracy and print it.

```python
# TODO: ADD YOUR CODE HERE
y_pred = model.predict(x_test)

accuracy = accuracy_score(y_test, y_pred)

print(f'Model Accuacy: {round(accuracy, 2)}')
```
